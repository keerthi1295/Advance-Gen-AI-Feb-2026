```python
#Problem 1: Unique Words in a Sentence

# Taking input
sentence = input("Enter sentence: ")

# Splitting sentence into words
words = sentence.split()

# Converting list to set to remove duplicates
unique_words = set(words)

print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)
```

```python
#Problem 2: Highest Salary from Employee Data

employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

# Finding employee with highest salary
highest_employee = max(employees, key=employees.get)
highest_salary = employees[highest_employee]

print("Highest Salary:", highest_employee, "-", highest_salary)
```

```python
#Problem 3: Find Maximum and Minimum Values

numbers = [45, 22, 89, 10, 66]

# Initializing max and min with first element
max_value = numbers[0]
min_value = numbers[0]

# Loop to compare each number
for num in numbers:
    if num > max_value:
        max_value = num
    if num < min_value:
        min_value = num

print("List:", numbers)
print("Max:", max_value)
print("Min:", min_value)
```

```python
#Problem 4: Count Products Above a Price Threshold

prices = [450, 1200, 899, 1500, 300]

count = 0

# Checking each price
for price in prices:
    if price > 1000:
        count += 1

print("Products above 1000:", count)
```

```python
#Problem 5: Calculate Attendance Percentage

attendance = ["P", "P", "A", "P", "P"]

# Counting Present days
present_count = attendance.count("P")

# Calculating percentage
percentage = (present_count / len(attendance)) * 100

print("Attendance Percentage:", percentage)
```

```python
# Problem 6: Remove Duplicate Phone Numbers

phone_numbers = [9876543210, 9123456789, 9876543210]

# Removing duplicates using set
unique_numbers = set(phone_numbers)

print("Unique phone numbers:", unique_numbers)
```

```python
#Problem 7: Count Character Frequency

string = "python"

frequency = {}

# Counting each character
for char in string:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
```

```python
# Problem 8: Convert List to Tuple

numbers = [10, 20, 30]

# Converting list to tuple
numbers_tuple = tuple(numbers)

print("Tuple:", numbers_tuple)
```

```python
# Problem 9: Check if a Key Exists in a Dictionary

employees = {
    "Ravi": 75000,
    "Anita": 68000
}

key_to_check = "Ravi"

# Checking if key exists
if key_to_check in employees:
    print("Employee exists")
else:
    print("Employee does not exist")
```

```python
# Problem 10: Calculate Average Marks

marks = [70, 75, 80, 65]

# Calculating average
average = sum(marks) / len(marks)

print("Average Marks:", average)
```
