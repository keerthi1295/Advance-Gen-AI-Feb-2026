```python
#Problem 1: Mobile Recharge Validation System

# Function to validate recharge amount
def validate_recharge(amount):
    valid_plans = [199, 299, 399, 599]

    if amount < 50:
        print("Recharge amount must be at least ₹50.")
        return False
    elif amount not in valid_plans:
        print("Invalid plan selected. Available plans: 199, 299, 399, 599")
        return False
    else:
        return True


# Main program with retry option
while True:
    try:
        amount = int(input("Enter recharge amount: ₹"))

        if validate_recharge(amount):
            print("Recharge Successful ✅")
            break
        else:
            print("Please try again.\n")

    except ValueError:
        print("Invalid input! Please enter a valid number.\n")
```

```python
#Problem 2: Inventory Reorder Alert System

# Function to check stock status
def check_inventory(products):
    for item, quantity in products.items():
        if quantity < 15:
            print(f"{item} → Reorder Alert (Stock: {quantity})")
        else:
            print(f"{item} → Stock OK (Stock: {quantity})")


# Sample dictionary
inventory = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 12,
    "Monitor": 18
}

check_inventory(inventory)
```

```python
#Problem 3: Student Result Processing System

def calculate_result(marks):
    total = 0
    for mark in marks:
        total = total + mark
    average = total / len(marks)

    if average >= 50:
        result = "Pass"
    else:
        result = "Fail"
    return average, result

students = {
    "Pavithra": [60, 70, 80],
    "Shivani": [40, 45, 50],
    "Teju": [55, 85, 85],
    "KRupa": [30, 35, 40]
}
for name, marks in students.items():

    avg, final_result = calculate_result(marks)

    print("Student Name:", name)
    print("Marks:", marks)
    print("Average:", avg)
    print("Result:", final_result)
    print("\n")
```

```python
#Problem 4: Cab Fare Estimator with Retry Option

# Function to calculate fare
def calculate_fare(distance, peak):
    base_fare = 50
    fare = base_fare + (12 * distance)

    if peak.lower() == "yes":
        fare += fare * 0.25

    return fare


# Retry option
while True:
    try:
        distance = float(input("Enter distance (km): "))
        peak = input("Is it peak hour? (yes/no): ")

        total_fare = calculate_fare(distance, peak)
        print(f"Total Fare: ₹{total_fare}")
        break

    except ValueError:
        print("Invalid input. Please try again.\n")
```

```python
#Problem 5: Employee Attendance Eligibility Checker

# Function to check eligibility
def check_eligibility(attendance):
    present_days = 0

    for day in attendance:
        if day == "P":
            present_days += 1

    percentage = (present_days / len(attendance)) * 100

    if percentage >= 75:
        return f"Attendance: {percentage}% → Eligible"
    else:
        return f"Attendance: {percentage}% → Not Eligible"


# Example
attendance_list = ["P", "A", "P", "P", "A", "P", "P"]
print(check_eligibility(attendance_list))
```

```python
#Problem 6: Password Strength Checker

# Function to check password strength
def check_password(password):
    has_digit = False
    has_special = False

    special_chars = "@#$"

    if len(password) < 8:
        return "Password too short ❌"

    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True

    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password"


# Example
pwd = input("Enter password: ")
print(check_password(pwd))
```
