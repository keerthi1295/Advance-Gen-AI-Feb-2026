```python

#Task 1. User Login Check

# Predefined credentials
username = 'admin'
password = '1234'

# Taking user input
input_username = input("Enter username: ")
input_password = input("Enter password: ")

# Checking if both username and password match
if input_username == username and input_password == password:
    print("Login Successful")
else:
    print("Invalid Credentials")
```

```python
#Task2. Pass / Fail Analyzer

# List of student marks
marks = [45, 78, 90, 33, 60]

# Initialize counters for pass and fail students
pass_count = 0
fail_count = 0

# Loop through each mark in the list
for mark in marks:

    # Check if the student passed (marks >= 50)
    if mark >= 50:
        pass_count += 1   # Increase pass count by 1
    else:
        fail_count += 1   # Increase fail count by 1

# Print total number of students
print("Total Students :", len(marks))

# Print number of students who passed
print("Pass Students  :", pass_count)

# Print number of students who failed
print("Fail Students  :", fail_count)
```

```python
#Task3. Simple Data Cleaner

# Original list of names
names = [" Alice ", "bob", " CHARLIE "]

# Create an empty list to store cleaned names
cleaned_names = []

# Loop through each name in the list
for name in names:

    # Remove leading and trailing spaces
    name = name.strip()

    # Convert the name to lowercase
    name = name.lower()

    # Add the cleaned name to the new list
    cleaned_names.append(name)

# Print the final cleaned list
print("Original Names:", names)
print("Cleaned Names:", cleaned_names)
```

```python
#Task4. Message Length Analyzer

# List of messages
messages = ["Hi", "Welcome to the platform", "OK"]

# Loop through each message in the list
for message in messages:

    # Calculate the length of the message using len()
    length = len(message)


    print("Message:", message)
    print("Length :", length)

    # Check if the message length is greater than 10
    if length > 10:
        print("Status : Long Message ⚠️")
    else:
        print("Status : Short Message")

    print()
```

```python
#Task 5. Error Message Detector

# system log entries
logs = ["INFO", "ERROR", "WARNING", "ERROR"]

# Initialize error counter
error_count = 0

# Count "ERROR" entries
for log in logs:
    if log == "ERROR":
        error_count += 1

# Print total error count
print("Total ERROR messages:", error_count)
```
