```python
# Problem 1: Social Media – Post Engagement Analyzer

def analyze_engagement(likes):
    total_likes = 0
    for like in likes:
        total_likes += like

    status = "Viral Post" if total_likes >= 1000 else "Normal Engagement"

    print("Total Likes:", total_likes)
    print("Post Status:", status)

# Example
analyze_engagement([200, 350, 500])
```

```python
# Problem 2: Healthcare – Medicine Stock Alert System

def check_medicine_stock(stock):
    status = "Low Stock Alert" if stock < 10 else "Stock Sufficient"

    print("Medicine Stock:", stock)
    print("Status:", status)

# Example
check_medicine_stock(6)
```

```python
# Problem 3: Agriculture – Rainfall Adequacy Checker

def rainfall_checker(rainfall, required_level):
    total = 0
    for r in rainfall:
        total += r

    average = total / len(rainfall)
    status = "Adequate Rainfall" if average >= required_level else "Inadequate Rainfall"

    print("Average Rainfall:", average)
    print("Rainfall Status:", status)

# Example
rainfall_checker([70, 75, 72, 71], 70)
```

```python
# Problem 4: Social Media – Duplicate Account Detection

def detect_duplicates(usernames):
    has_duplicates = len(usernames) != len(set(usernames))

    print("Duplicate Accounts Found:", "Yes" if has_duplicates else "No")

# Example
detect_duplicates(["john", "alex", "john"])
```

```python
# Problem 5: Healthcare – Appointment Eligibility Checker

def check_eligibility(age):
    status = "Eligible" if age >= 18 else "Not Eligible"

    print("Patient Age:", age)
    print("Eligibility Status:", status)

# Example
check_eligibility(21)
```

```python
# Problem 6: Agriculture – Premium Crop Price Filter

def premium_crop_filter(prices):
    premium = []
    for price in prices:
        if price > 2000:
            premium.append(price)

    print("Premium Crops:", premium)

# Example
premium_crop_filter([1800, 2500, 3200, 1500])
```

```python
# Problem 7: System Monitoring – Application Health Checker

def app_health_checker(errors):
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print("Error Count:", errors)
    print("System Status:", status)

# Example
app_health_checker(7)
```

```python
# Problem 8: Banking – Daily Transaction Limit Checker

def transaction_checker(amount):
    limit = 50000
    status = "Approved" if amount <= limit else "Rejected"

    print("Transaction Amount:", amount)
    print("Transaction Status:", status)

# Example
transaction_checker(60000)
```

```python
# Problem 9: E-Learning – Student Attendance Eligibility System

def attendance_checker(attendance):
    total_classes = len(attendance)
    attended = 0

    for day in attendance:
        if day == 1:
            attended += 1

    percentage = (attended / total_classes) * 100
    status = "Eligible" if percentage >= 75 else "Not Eligible"

    print("Attendance Percentage:", percentage)
    print("Exam Eligibility:", status)

# Example
attendance_checker([1, 1, 1, 0, 1])
```

```python
# Problem 10: Smart Electricity Bill Analyzer

def calculate_electricity_bill(units):
    """
    Calculates electricity bill based on slab system
    and classifies usage level.
    """

    bill = 0

    # Slab calculation
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    # Usage classification
    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"

    return bill, status


units_consumed = 250

total_bill, usage_status = calculate_electricity_bill(units_consumed)

print("Units Consumed:", units_consumed)
print("Total Bill:", total_bill)
print("Usage Level:", usage_status)
```
