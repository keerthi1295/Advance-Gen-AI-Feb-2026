```python
# Problem 1: Smart Parking Lot Management System
def parking_lot_manager(capacity, logs):
    parked = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1

    status = "Available"
    if parked >= capacity:
        status = "Full"

    print("Currently Parked Vehicles:", parked)
    print("Parking Status:", status)


# Function call (THIS IS REQUIRED)
parking_lot_manager(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
# Problem 2: Online Food Delivery Time Estimator

def estimate_delivery_time(distance, traffic, weather):
    time = distance * 5  # base time: 5 minutes per km

    if traffic == "High":
        time += 15
    elif traffic == "Medium":
        time += 8

    if weather == "Rainy":
        time += 10
    elif weather == "Stormy":
        time += 20

    print("Estimated Delivery Time:", time, "minutes")
    print()


# FUNCTION CALL (THIS IS REQUIRED)
estimate_delivery_time(8, "High", "Rainy")
```

```python
# Problem 3: Movie Theatre Seat Occupancy Analyzer

def seat_occupancy_analyzer(total_seats, booked_seats):
    # Count booked seats
    booked_count = len(booked_seats)

    # Calculate occupancy percentage
    occupancy = (booked_count / total_seats) * 100

    # Determine show status
    if occupancy == 100:
        status = "Housefull"
        suggestion = "No additional shows required"
    elif occupancy >= 70:
        status = "Almost Full"
        suggestion = "Consider opening additional shows"
    else:
        status = "Seats Available"
        suggestion = "No need for additional shows"

    # Display results
    print("Occupancy:", int(occupancy), "%")
    print("Show Status:", status)
    print("Suggestion:", suggestion)


# FUNCTION CALL (IMPORTANT)
seat_occupancy_analyzer(200, [1] * 150)
```

```python
# Problem 4: Cloud Server Load Classification System

def server_load_classifier(cpu_readings):
    total = 0

    # Loop to calculate total CPU usage
    for cpu in cpu_readings:
        total += cpu

    # Calculate average CPU load
    average_cpu = total / len(cpu_readings)

    # Classify server status
    if average_cpu < 50:
        status = "Normal"
    elif average_cpu <= 80:
        status = "Warning"
    else:
        status = "Critical"

    # Display results
    print("Average CPU Load:", int(average_cpu), "%")
    print("Server Status:", status)


# FUNCTION CALL (REQUIRED TO GET OUTPUT)
server_load_classifier([45, 60, 70, 85, 90])
```

```python
# Problem 5: Smart Classroom Resource Usage Monitor

def resource_usage_monitor(resource_usage):
    overused_resources = []

    # Check each resource usage
    for resource, hours in resource_usage.items():
        if hours > 8:   # condition for overuse
            overused_resources.append(resource)

    # Display results
    if overused_resources:
        print("Overused Resources:", ", ".join(overused_resources))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# FUNCTION CALL (REQUIRED TO GET OUTPUT)
resource_usage_monitor({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

```python
# Problem 6: Online Event Registration Capacity Controller

def event_registration_controller(capacity, registrations):
    confirmed = 0
    waitlisted = 0

    # Loop to process each registration
    for i in range(registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1

    # Display results
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)

    if waitlisted > 0:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")


# FUNCTION CALL (REQUIRED TO GET OUTPUT)
event_registration_controller(100, 105)
```
